```python
cust_seg['r20']=cust_seg['recency'].quantile(0.2)
cust_seg['r40']=cust_seg['recency'].quantile(0.4)
cust_seg['r60']=cust_seg['recency'].quantile(0.6)
cust_seg['r80']=cust_seg['recency'].quantile(0.8)
cust_seg['r100']=cust_seg['recency'].quantile(1)
cust_seg['f20']=cust_seg['frequency'].quantile(0.2)
cust_seg['f40']=cust_seg['frequency'].quantile(0.4)
cust_seg['f60']=cust_seg['frequency'].quantile(0.6)
cust_seg['f80']=cust_seg['frequency'].quantile(0.8)
cust_seg['f100']=cust_seg['frequency'].quantile(1)
```

```python
cust_seg
```

```python
def func1(x):
  if x['monetary']<=x['m20']:
    x['m_score']=1
  elif x['monetary']<=x['m40'] and x['monetary']>x['m20']:
    x['m_score']=2
  elif x['monetary']<=x['m60'] and x['monetary']>x['m40']:
    x['m_score']=3
  elif x['monetary']<=x['m80'] and x['monetary']>x['m60']:
    x['m_score']=4
  else:
    x['m_score']=5
  return x

```

```python
def func2(x):
  if x['recency']<=x['r20']:
    x['r_score']=5
  elif x['recency']<=x['r40'] and x['recency']>x['r20']:
    x['r_score']=4
  elif x['recency']<=x['r60'] and x['recency']>x['r40']:
    x['r_score']=3
  elif x['recency']<=x['r80'] and x['recency']>x['r60']:
    x['r_score']=2
  else:
    x['r_score']=1
  return x
```

```python
def func3(x):
  if x['frequency']<=x['f20']:
    x['f_score']=1
  elif x['frequency']<=x['f40'] and x['frequency']>x['f20']:
    x['f_score']=2
  elif x['frequency']<=x['f60'] and x['frequency']>x['f40']:
    x['f_score']=3
  elif x['frequency']<=x['f80'] and x['frequency']>x['f60']:
    x['f_score']=4
  else:
    x['f_score']=5
  return x
```

```python
cust_seg1=cust_seg.apply(func2,axis=1)
```

```python
cust_seg1
```

```python
cust_seg2=cust_seg1.apply(func1,axis=1)
```

```python
cust_seg3=cust_seg2.apply(func3,axis=1)
```

```python
cust_seg3
```

```python
cust_seg3['fm_score']=round((cust_seg3['f_score']+cust_seg3['m_score'])/2,0)
```

```python
cust_seg3
```

```python
def func4(x):
  if (x['r_score']==5 and x['fm_score']==5) or (x['r_score']==5 and x['fm_score']==4) or (x['r_score']==4 and x['fm_score']==5):
    x['rfm_segment']='Champions'
  elif (x['r_score']==5 and x['fm_score']==3) or (x['r_score']==4 and x['fm_score']==4) or (x['r_score']==3 and x['fm_score']==5) or (x['r_score']==3 and x['fm_score']==4):
    x['rfm_segment']='Loyal Customers'
  elif (x['r_score']==5 and x['fm_score']==2) or (x['r_score']==4 and x['fm_score']==2) or (x['r_score']==3 and x['fm_score']==3) or (x['r_score']==4 and x['fm_score']==3):
    x['rfm_segment']='Potential Loyalists'
  elif x['r_score']==5 and x['fm_score']==1:
    x['rfm_segment']='Recent Customers'
  elif (x['r_score']==4 and x['fm_score']==1) or (x['r_score']==3 and x['fm_score']==1):
    x['rfm_segment']='Promising'
  elif (x['r_score']==3 and x['fm_score']==2) or (x['r_score']==2 and x['fm_score']==3) or (x['r_score']==2 and x['fm_score']==2):
    x['rfm_segment']='Customer Needing Attention'
  elif (x['r_score']==2 and x['fm_score']==1):
    x['rfm_segment']='About to Sleep'
  elif (x['r_score']==2 and x['fm_score']==5) or (x['r_score']==2 and x['fm_score']==4) or (x['r_score']==1 and x['fm_score']==3):
    x['rfm_segment']='At Risk'
  elif (x['r_score']==1 and x['fm_score']==5) or (x['r_score']==1 and x['fm_score']==4):
    x['rfm_segment']='Cant Loose Them'
  elif (x['r_score']==1 and x['fm_score']==2):
    x['rfm_segment']='Hibernating'
  elif (x['r_score']==1 and x['fm_score']==1):
    x['rfm_segment']='Lost'
  return x
```
